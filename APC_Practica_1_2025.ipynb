```python
le_sex = LabelEncoder()
le_embarked = LabelEncoder()
#pasem de dades categoriques a numeriques
dftrain["Sex"] = le_sex.fit_transform(dftrain["Sex"])
dftest["Sex"] = le_sex.transform(dftest["Sex"])

dftrain["Embarked"] = le_embarked.fit_transform(dftrain["Embarked"])
dftest["Embarked"] = le_embarked.transform(dftest["Embarked"])

#normalitzacio
scaler = StandardScaler()
num_vars = ["Age", "Fare", "SibSp", "Parch"]
dftrain[num_vars] = scaler.fit_transform(dftrain[num_vars])
dftest[num_vars] = scaler.transform(dftest[num_vars])


#df final

X = dftrain[["Sex","PassengerId", "Pclass", "Age", "SibSp", "Parch", "Fare", "Embarked"]]
y = dftrain["Survived"]

print(X.head())
corr_with_target = X.corrwith(y).sort_values(ascending=False)
print(corr_with_target)

#treiem passengerId perque no aporta res.
X = dftrain[["Sex", "Pclass", "Age", "SibSp", "Parch", "Fare", "Embarked"]]
```

```python
def avaluar_model(model, X, y, test_size=0.4, random_state=42):
    print(f"\nModel: {model.__class__.__name__}")
    
    # Separar train/test
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state, stratify=y
    )
    
    # Entrenament
    model.fit(X_train, y_train)
    
    # Prediccions
    y_scores = model.predict_proba(X_test)[:, 1]
    
    y_pred = (y_scores >= 0.5).astype(int)

    # Mètriques
    print("=== Classification Report ===")
    print(classification_report(y_test, y_pred))
    
    f1 = f1_score(y_test, y_pred)
    acc = accuracy_score(y_test, y_pred)
    print(f"F1-score global: {f1:.3f}")
    print(f"Accuracy: {acc:.3f}")
    

    # Precision-Recall Curve 
    precision, recall, _ = precision_recall_curve(y_test, y_scores)
    avg_precision = average_precision_score(y_test, y_scores)
    plt.figure(figsize=(6,5))
    plt.plot(recall, precision, color='blue', lw=2, label=f'AP = {avg_precision:.3f}')
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    plt.title('Precision–Recall Curve')
    plt.legend(loc='lower left')
    plt.grid(True)
    plt.tight_layout()
    plt.show()


    # ROC Curve 
    fpr, tpr, _ = roc_curve(y_test, y_scores)
    roc_auc = roc_auc_score(y_test, y_scores)
    plt.figure(figsize=(6,5))
    plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'AUC = {roc_auc:.3f}')
    plt.plot([0, 1], [0, 1], color='navy', lw=1, linestyle='--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('ROC Curve')
    plt.legend(loc='lower right')
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    # Mètriques numèriques
    print(f"Average Precision (AUC-PR): {avg_precision:.3f}")
    print(f"AUC-ROC: {roc_auc:.3f}")


    # Matriu de confusió 
    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm)
    disp.plot(cmap='Blues')
    plt.title("Matriu de confusió")
    plt.tight_layout()
    plt.show()


```

```python
# Models
model1 = LogisticRegression()
model2 = DecisionTreeClassifier()
model3 = RandomForestClassifier()
model4 = GradientBoostingClassifier()

# Crides
avaluar_model(model1, X, y)
avaluar_model(model2, X, y)
avaluar_model(model3, X, y)
avaluar_model(model4, X, y)
```

```python
def cross_validation(X, y, cv_folds=5, test_size=0.4):
    # Dividim les dades en train i validació
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=42, stratify=y
    )

    models = {
        "Regressió logística": LogisticRegression(),
        "Arbre de decisió": DecisionTreeClassifier(),
        "Random Forest": RandomForestClassifier(),
        "Gradient Boosting": GradientBoostingClassifier()
    }

    results = []
    cv = KFold(n_splits=cv_folds, shuffle=True, random_state=42)
    scorer = make_scorer(f1_score)

    for name, model in models.items():
        print(f"\nModel: {name}")
        start = time.time()

        # VALIDACIÓ CREUADA dins del train
        cv_scores = cross_val_score(model, X_train, y_train, scoring=scorer, cv=cv, n_jobs=-1)
        cv_mean, cv_std = cv_scores.mean(), cv_scores.std()
        print(f"F1-CV (mean ± std): {cv_mean:.3f} ± {cv_std:.3f}")

        # ENTRENAMENT FINAL i avaluació sobre validació
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        f1_val = f1_score(y_test, y_pred)

        elapsed = round(time.time() - start, 2)
        print(f"F1-validació: {f1_val:.3f} | Temps: {elapsed}s")

        results.append({
            "Model": name,
            "F1-CV-mean": round(cv_mean, 3),
            "F1-CV-std": round(cv_std, 3),
            "F1-validació": round(f1_val, 3),
            "Temps (s)": elapsed
        })

    # DataFrame resum ordenat
    results_df = pd.DataFrame(results).sort_values(by="F1-validació", ascending=False).reset_index(drop=True)
    print("\n=== Resum comparatiu ===")
    print(results_df)
```

```python
def optimitzar_models(X, y, n_iter=20, test_size=0.2, random_state=42):
    random.seed(random_state)
    np.random.seed(random_state)

    # Definim grids
    grids = {
        "Regressió logística": {
            "C": [0.01, 0.1, 1, 10],
            "penalty": ["l1", "l2"],
            "class_weight": [None, "balanced"]
        },
        "Arbre de decisió": {
            "max_depth": [3, 5, 7, 10, None],
            "min_samples_split": [2, 4, 6],
            "min_samples_leaf": [1, 2, 4],
            "class_weight": [None, "balanced"]
        },
        "Random Forest": {
            "n_estimators": [50, 100, 200],
            "max_depth": [5, 8, 10, None],
            "min_samples_split": [2, 4, 6],
            "min_samples_leaf": [1, 2, 4],
            "max_features": ["sqrt", "log2", None],
            "class_weight": [None, "balanced"]
        },
        "Gradient Boosting": {
            "n_estimators": [100, 200],
            "learning_rate": [0.01, 0.05, 0.1],
            "max_depth": [3, 4, 5],
            "min_samples_split": [2, 4, 6],
            "subsample": [0.8, 1.0]
        }
    }


    model_classes = {
        "Regressió logística": LogisticRegression,
        "Arbre de decisió": DecisionTreeClassifier,
        "Random Forest": RandomForestClassifier,
        "Gradient Boosting": GradientBoostingClassifier
    }

    best_models = {}

    for name in grids.keys():
        print(f"\nOptimitzant {name} amb cerca manual...")
        grid = grids[name]

        # Generar combinacions
        keys, values = zip(*grid.items())
        all_combinations = [dict(zip(keys, v)) for v in itertools.product(*values)]

        # Si n_iter < total combinacions, mostrem una mostra aleatòria
        if n_iter < len(all_combinations):
            all_combinations = random.sample(all_combinations, n_iter)

        best_f1 = -1
        best_params = None
        best_model = None

        for params in all_combinations:
            # Crear model amb aquests params
            if name == "Regressió logística":
                model = LogisticRegression(solver='liblinear', random_state=random_state, **params)
            else:
                model = model_classes[name](random_state=random_state, **params)

            # Aplicar cross_validation amb 1 fold (train/test)
            X_train, X_test, y_train, y_test = train_test_split(
                X, y, test_size=test_size, stratify=y, random_state=random_state
            )
            model.fit(X_train, y_train)
            y_pred = model.predict(X_test)
            f1 = f1_score(y_test, y_pred)

            if f1 > best_f1:
                best_f1 = f1
                best_params = params
                best_model = model

        print(f"Millors paràmetres: {best_params}")
        print(f"F1-test amb aquests params: {best_f1:.3f}")
        best_models[name] = best_model

    return best_models
```

```python
def cross_validation(X, y, cv_folds=5, test_size=0.2):
    # Dividim les dades en train i validació
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=42, stratify=y
    )

    # Definim els models amb els hiperparàmetres ja optimitzats
    models = {
        "Regressió logística": LogisticRegression(C=10, penalty='l1', class_weight='balanced', solver='liblinear', random_state=42),
        "Arbre de decisió": DecisionTreeClassifier(max_depth=None, min_samples_split=6, min_samples_leaf=1, class_weight=None, random_state=42),
        "Random Forest": RandomForestClassifier(n_estimators=200, max_depth=10, max_features='log2', min_samples_split=6, min_samples_leaf=4, class_weight='balanced', random_state=42),
        "Gradient Boosting": GradientBoostingClassifier(n_estimators=100, learning_rate=0.05, max_depth=5, min_samples_split=6, subsample=0.8, random_state=42)
    }

    results = []
    cv = KFold(n_splits=cv_folds, shuffle=True, random_state=42)
    scorer = make_scorer(f1_score)

    for name, model in models.items():
        print(f"\nModel: {name}")
        start = time.time()

        # VALIDACIÓ CREUADA dins del train
        cv_scores = cross_val_score(model, X_train, y_train, scoring=scorer, cv=cv, n_jobs=-1)
        cv_mean, cv_std = cv_scores.mean(), cv_scores.std()
        print(f"F1-CV (mean ± std): {cv_mean:.3f} ± {cv_std:.3f}")

        # ENTRENAMENT FINAL i avaluació sobre validació
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        f1_val = f1_score(y_test, y_pred)

        elapsed = round(time.time() - start, 2)
        print(f"F1-validació: {f1_val:.3f} | Temps: {elapsed}s")

        results.append({
            "Model": name,
            "F1-CV-mean": round(cv_mean, 3),
            "F1-CV-std": round(cv_std, 3),
            "F1-validació": round(f1_val, 3),
            "Temps (s)": elapsed
        })

    # DataFrame resum ordenat
    results_df = pd.DataFrame(results).sort_values(by="F1-validació", ascending=False).reset_index(drop=True)
    print("\n=== Resum comparatiu ===")
    print(results_df)
```
